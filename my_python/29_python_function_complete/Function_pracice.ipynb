```python
def fib(n):
    a = 0
    b = 1
    
    print(a)
    print(b)

    for i in range(0,n):
        c = a + b
        a = b
        b = c

        print(c)
fib(10)
```

```python
def fib(n):
    a, b = 0, 1
    if n == 1:
        print(a)
    else:
        print(a)
        print(b)

        for i in range(2, n):
            c = a + b
            a = b
            b = c
            print(c)
fib(5)
```

# Fctorial of a Number in Python

```python
def fact(n):
    f = 1
    for i in range(1, n+1):
        f = f*i
    return f
x = 4
result = fact(x)
print(result)
```

# Python decorator

```python
def div(a,b):
    print(a//b)
div(4,2)
```

```python
def div(a, b):
    print(a // b)
div(2,4) 
```

```python
def div(a,b):
    if a<b:
        a,b = b,a
    print(a//b)
div(2,4)
```

```python
def div(a,b):
    print(a / b)

def div_decorator(func): # hear div_dectorator will accept the div function
    def inner(a,b):
        if a<b:
            a,b = b,a
        return func(a,b)
    return inner

div = div_decorator(div)

div(2,4)   
```
