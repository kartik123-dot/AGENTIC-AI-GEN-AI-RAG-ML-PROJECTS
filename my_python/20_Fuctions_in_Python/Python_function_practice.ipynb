```python
def add_sub_mul(x,y):
    c = x + y
    d = x - y
    e = x * y
    return c,d,e
r, r1, r2 = add_sub_mul(7,8)

print(r)
print(r1)
print(r2)
```

# 1-Positional argument

```python
def person(name, age):
    print(name)
    print(age)
person('nit')
```

```python
def person(name, age):
    print(name)
    print(age)
person(23)
```

```python
def person(name, age):
    print(name)
    print(age)
person('nit', 23, 45)
```

```python
def person(name, age):
    print(name)
    print(age)
person('nit', 23)
```

```python
def person(name, age):
    print(name)
    print(age+3)
person(23, 'nit')
```

```python
def person(name, age):
    print(name)
    print(age+3)
person('nit',23)
```

# Keyword argument

```python
def person(name, age):
    print(name)
    print(age+1)
person(name = 'Kartik', age = 23)
```

```python
def person(name, age, mob):
    print(name)
    print(age+1)
    print(mob)
person(name = 'Kartik', age = 23, 123456)
```

```python
def person(name, age, mob):
    print(name)
    print(age+1)
    print(mob)
person(name = 'Kartik', age = 23, mob=123456)
```

# Default aargument

```python
def person(name, age):    
    print(name)
    print(age) 

person('vote') 
```

```python
def person(name, age = 18):    
    print(name)
    print(age) 

person('vote') 
```

```python
def person(name, age = 18):    
    print(name)
    print(age - 1) 

person('vote') 
```
